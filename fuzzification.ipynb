make a model to implement that a input came and convert into fuzzy number by using formula.

```python
temperature = int(input("Enter the temperature :"))
#1. fuzzfication- convert crisp into fuzzy
#low
if temperature<=20:
  low=1
elif 20<temperature<30:#low+high
  low=(30*temperature)/30-20
else:
  low=0

#medium
if temperature<=20 or temperature>=50:
  mid=0
elif 20<temperature<30 :#low+mid
  mid=(20*temperature)/30-20
elif 30<temperature<=40:
  mid=1
else :#mid+high
  mid=(50*temperature)/50-40

#high
if temperature<40:
  high=0
elif 40<temperature<=50:
  high=(50*temperature)/50-40# mid+high
else:
  high=1

#rules
slow = low
moderate = mid
fast = high

#Defuzzification
num = slow*25+moderate*50+fast*75
den=slow+moderate+fast

if den ==0:
  fan_speed=0
else:
  fan_speed=num/den

print("temperature=", temperature,"C")
print("low Membership=",round(low,2))
print("medium Membership=",round(mid,2))
print("high Membership=",round(high,2))
print("fan Speed=",round(fan_speed,2),"%")

if fan_speed<=30:
  print("fan Status:SLOW")
elif fan_speed<=60:
  print("fan Status:MEDIUM ")
else:
  print("fan Status:HIGH")
```
